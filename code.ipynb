```python
import pandas as pd
from pathlib import Path

csv_path = Path(r"C:\Users\User\AI-Career-Analyzer\hellojob_150_sehife.csv")

# 1. sep="\t" parametrini silirik (standart vergül ayırıcısı işləsin)
df = pd.read_csv(csv_path)

# 2. "Vakansiya adı" sütunundakı \n və "PREMİUM" sözlərini təmizləyirik
df["Vakansiya adı"] = df["Vakansiya adı"].astype(str).apply(lambda x: x.split('\n')[0].replace("PREMİUM", "").strip())

# 3. Şirkət adındakı gərəksiz boşluqları təmizləyirik
df["Şirkət"] = df["Şirkət"].astype(str).str.strip()

df.head(10)
```

```python
import pandas as pd

# 1. İstifadəçinin CV-sindən çıxarılan bacarıqlar (NLP vasitəsilə)
user_skills = {"python", "pandas", "sql", "git", "data analysis"}

# 2. Scraping etdiyiniz vakansiya bazasından nümunə
jobs_data = [
    {
        "Vakansiya": "Data Analyst",
        "Sirket": "Kapital Bank",
        "Required_Skills": {"python", "sql", "pandas", "tableau", "excel", "power bi"},
        "Link": "https://www.hellojob.az/..."
    },
    {
        "Vakansiya": "Machine Learning Engineer",
        "Sirket": "PASHA Bank",
        "Required_Skills": {"python", "sql", "pandas", "scikit-learn", "xgboost", "docker", "git"},
        "Link": "https://www.hellojob.az/..."
    }
]

# 3. Uyğunluq və Çatışmayan Bacarıqların (Skill Gap) hesablanması
results = []

for job in jobs_data:
    required = job["Required_Skills"]
    
    # Uyğun gələn bacarıqlar
    matched_skills = user_skills.intersection(required)
    
    # Çatışmayan bacarıqlar - Öyrənilməli olanlar (Skill Gap)
    missing_skills = required.difference(user_skills)
    
    # Uyğunluq faizi
    match_score = round((len(matched_skills) / len(required)) * 100, 2)
    
    results.append({
        "Vakansiya": job["Vakansiya"],
        "Şirkət": job["Sirket"],
        "Uyğunluq Faizi": f"{match_score}%",
        "Bildiyiniz Bacarıqlar": list(matched_skills),
        "Gələcəkdə Öyrənilməli (Skill Gap)": list(missing_skills),
        "Link": job["Link"]
    })

# Nəticələri cədvəl şəklində sıralayırıq
df_results = pd.DataFrame(results)
df_results = df_results.sort_values(by="Uyğunluq Faizi", ascending=False)

print(df_results.to_string())
```
